```python
import os
import sys

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import json
from copy import deepcopy

import pickle
import pandas as pd
from datetime import datetime, timedelta
import numpy as np

from stable_platform_matchings import Optimizer
from stable_platform_matchings.optimization.options import OptimizerParams, SolverOptions
from stable_platform_matchings.domain.instance import Instance
from stable_platform_matchings.domain.entities import Farmer, Intermediary, Mill
from stable_platform_matchings.graphs.road_graphs import RoadGraph

sys.path.insert(0, '../../FactoredPlatformSolver/src')
from farmers_intermediaries import Instance as OldInstance
from road_graphs import RoadGraph as OldRoadGraph
import pickle
import pandas as pd
import pricing as OldPricing
from pricing import Optimizer as OldOptimizer

new_json_files = {}

new_directory = "../results/results_4"

for filename in os.listdir(new_directory):
    if filename.endswith(".json"):
        filepath = os.path.join(new_directory, filename)
        with open(filepath, 'r') as file:
            new_json_files[filename] = json.load(file)
```

```python
data = new_json_files["100.json"][0]
instance_str = data["instance_str"]
quantities = data["farmer_quantities"]

old_instance = OldInstance.from_yaml(f"../../FactoredPlatformSolver/data/instances/{instance_str}.yaml", opt_quantities=quantities)

with open("../data/graph_0-14960_00_new.pickle", 'rb') as pickle_file:
    G = pickle.load(pickle_file)

old_instance.set_graph(OldRoadGraph(G))
```

```python
het_costs = data["cost"]
epsilons = data["epsilon"]

parameters = {
        "epsilon":epsilons, 
        "solver": "gurobi", 
        "het_costs": het_costs,
    }

opt = OldOptimizer(old_instance, parameters)


summary_vanilla = opt.solve("heuristic_optimized", options={
        "structured_farmer_prices": False,
        "domination": False,})
```

```python
instance_str = data["instance_str"]
quantities = data["farmer_quantities"]

instance = Instance.from_yaml(f"../data/reformatted_instances/{instance_str}.yaml", force_quantities=quantities)


with open("../data/graph_0-14960_00_new.pickle", 'rb') as pickle_file:
    G = pickle.load(pickle_file)

instance.set_graph(RoadGraph(G))
```

```python
het_costs = data["cost"]
epsilons = data["epsilon"]

options = SolverOptions(aggregate=True)
params = OptimizerParams(het_costs, epsilons, threads=14)
```
